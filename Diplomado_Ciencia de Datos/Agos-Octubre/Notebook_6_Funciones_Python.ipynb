# <FONT SIZE=5 COLOR="purple"> 1. Funciones en Python </FONT>

- Las funciones son bloques de código reutilizables que permiten organizar y estructurar programas de manera eficiente.

- Su principal objetivo es encapsular tareas específicas, facilitando la lectura, el mantenimiento y la reutilización del código.

- En Python, las funciones se definen mediante la palabra clave *def*
y pueden recibir parámetros de entrada, realizar operaciones y devolver resultados.

```python
def function(x,y,z..):
  .....
  return ...  
```
- El uso de funciones contribuye a reducir la duplicación de código, mejorar la modularidad de los programas y simplificar la resolución de problemas complejos mediante su división en tareas más pequeñas.

- Favorecen la claridad del código y permiten desarrollar aplicaciones más robustas y fáciles de depurar.

## <FONT SIZE=4 COLOR="purple"> 1.1 Observación sobre print y return en una función </FONT>

````python
def funcion2(x,y):
  z = x**2+y**2+1
  print(z)
````

````python
def funcion3(x,y):
  z = x**2+y**2+1
  return(z)
````

````python
funcion2(2,6)
````

````python
funcion3(2,6)
````

Los resultados son los mismos. Entonces ¿cuál es la diferencia?

````python
x = funcion2(2,6)
x+10
````

````python
y = funcion3(2,6)
y+10
````

La diferencia principal es que `print()` muestra información en pantalla, mientras que `return` devuelve un valor para que pueda ser utilizado por el resto del programa.

Podemos pensar como una calculadora.

- **print**:

  - la calculadora muestra el resultado en la pantalla.
  - cuando se quiere mostrar únicamente la información

- **return**:

  - la calculadora te entrega el resultado para que puedas usarlo en otro cálculo.
  - cuando queremos un resultados para seguir trabajando con él.



## <FONT SIZE=4 COLOR="purple"> 1.2 Formas diferentes del `print()` </FONT>

Para imprimir variables dentro del `print()` hay varias alternativas

- **f-strings**

- **.format()**

### <FONT SIZE=4 COLOR="purple"> 1.2.1 Forma tradicional </FONT>

Colocar la variable dentro del `print()` pero afuera de las comillas

````python
# primera forma
def saludo3(nombre):
  print("Hola", nombre, "espero que estés bien")
````

````python
saludo3("Carolina")
````

### <FONT SIZE=4 COLOR="purple"> 1.2.2 Método `f"{variable}"` </FONT>

Colocar la variable dentro de ls comillas y que Python no lo tome como cadena de texto.



````python
# método f"{variable}"
def saludo4(nombre):
  print(f"Buenos días {nombre}, espero que estés bien")
````

````python
saludo4("Ana María")
````

### <FONT SIZE=4 COLOR="purple"> 1.2.3 Método `"{}".format(variable)` </FONT>

También permite escribir la variable dentro de las comillas y que Python la identifique como variable y no cadena de texto.

````python
def saludo5(nombre):
  print("Hola {}, me alegra mucho verte".format(nombre))
````

````python
saludo5("Carlos")
````

**Ejemplo 3**

````python
def exploracion(datos):
    print("Primeros 5 registros \n")
    display(datos.head())

    print("\n Últimos 5 registros \n")
    display(datos.tail())

    print("\n Tamaño de los datos \n")
    print(datos.shape)
    print(f"Filas: {datos.shape[0]}")
    print(f"Columnas: {datos.shape[1]}")

    print("\n Nombres de las columnas \n")
    print(datos.columns)

    print("\n Información de los datos \n")
    print(datos.info())

    print("\n Estadísticas descriptivas \n")
    display(datos.describe())

    print("\n Valores faltantes \n")
    display(datos.isna().sum())
````

````python
exploracion(credito)
````

**Ejemplo 6**

````python
# detectar todas las filas donde esta variable no tiene un valor numérico válido.
def problema_texto(datos, variable):
  u = pd.to_numeric(datos[variable], errors='coerce').isna()
  posiciones = np.where(u)[0]
  print(posiciones)
  return datos.iloc[posiciones]
````

````python
problema_texto(credito,"Meses de Historial")
````

**Ejemplo 7**

````python
# ¿qué hace la siguiente función?
def fuera_rango(datos, variable, minimo, maximo):
    problema = (datos[variable] < minimo) | (datos[variable] > maximo)
    posiciones = np.where(problema)[0]
    print("Posiciones:", posiciones)
    return datos.iloc[posiciones]
````

````python
fuera_rango(homicidios, "Edad", 10, 80)
````

````python
def revisar_na(datos):
  resultado = pd.DataFrame({"cantidad_na": datos.isna().sum(),
                            "porcentaje_na": datos.isna().mean() * 100})
  return resultado
````
